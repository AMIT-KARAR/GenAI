```python
##Tokenization
## Paragraph --> Sentence

##  Tokenization
## Sentence-->paragraphs
from nltk.tokenize import sent_tokenize
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
documents=sent_tokenize(corpus)

type (documents)

print (documents)
```

```python
for sentance in documents:
  print (sentance)
```
